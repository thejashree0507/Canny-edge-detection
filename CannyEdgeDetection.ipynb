## Canny edge detection

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
img=cv2.imread("dog.jpg")
```

```python
img_gray=cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)
```

```python
blur=cv2.GaussianBlur(img,(5,5),1.4)
```

```python
edges=cv2.Canny(blur,100,100)
```

```python
plt.figure(figsize = (12,16))
plt.subplot(1,2,1);plt.axis("off");plt.imshow(img[:,:,::-1]);plt.title('Original') 
plt.subplot(1,2,2);plt.axis("off");plt.imshow(img_gray,cmap='gray');plt.title('Grayscale')
```

```python
plt.figure(figsize = (12,16))
plt.subplot(1,2,1);plt.axis("off");plt.imshow(blur[:,:,::-1]);plt.title('Blur image')
plt.subplot(1,2,2);plt.axis("off");plt.imshow(edges,cmap='gray');plt.title('Canny edge detection')
```
